<h4><b>SEMANTIC SIMILARITY WITH EMBEDDINGS</b></h4></span>

```python
sentences = [
    "The king is the ruler of the country.",
    "The queen governs her people with wisdom.",
    "I love a good pepperoni pizza.",
    "Monarchy is a form of government."
]
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
embeddings = model.encode(sentences)
```

```python
embeddings
```

```python
embeddings[0] # The king is the ruler of the country.
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarities = cosine_similarity([embeddings[0]], embeddings[1:])
```

```python
for i, score in enumerate(similarities[0]):
    print(sentences[i + 1], "->", round(float(score), 4))
```

```python
best = similarities[0].argmax() + 1
print("Most similar sentence:", sentences[best])
print("Score:", similarities[0].max())
```
