# Question 2: Control Structures

### **a) Write a program that takes an integer input from the user and prints whether it is even or odd.**

```python
def even_odd_checker():
    # Take an integer from the user
    number = int(input("Enter a number: "))

    # A number divisible by 2 with no remainder is even
    if number % 2 == 0:
        print(f"{number} is Even")
    else:
        print(f"{number} is Odd")

even_odd_checker()
```

```python
even_odd_checker()
```

### **b) Implement a program that checks whether a number is positive, negative, or zero.**

```python
def describe_num():
    user_input = int(input("Enter a number: "))

    if user_input == 0:
        print("Zero: it is neither positive nor negative")
    elif user_input < 0:
        print("Negative integer")
    else:
        print("Positive integer")

describe_num()
```

```python
describe_num()
```
