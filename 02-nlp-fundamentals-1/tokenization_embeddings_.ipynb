```python
import requests

URL = "https://quotes.toscrape.com/"
RESPONSE = requests.get(URL)
print(RESPONSE.status_code)
```

CONVERTING TEXT TO LOWERCASE

```python
import string
print(string.punctuation)
Data = Data.str.translate(str.maketrans("", "", string.punctuation))
```

<span style="color:green"> 
REPLACING NUMERIC VALUES WITH WORDS WHILE PRESERVING THEIR SEMANTIC MEANING
</span>

```python
! pip install num2words
```

```python
import re
from num2words import num2words

def replace_numbers(text):
    return re.sub(
        r"\d+",
        lambda x: num2words(int(x.group())),
        text
    )

text = "I have 2 dogs and 150 rupees."
print(replace_numbers(text))
```

```python
Data = Data.apply(replace_numbers)
```

<span style="color:green"> 
STOPWORD REMOVAL by Grammatical role
</span>

```python
nltk.download("averaged_perceptron_tagger_eng")
```

```python
from nltk.corpus import stopwords
nltk.download("stopwords")

# NLTK English stopwords, but keep negations so meaning is preserved
negations = {"no", "nor", "not"}
stop_words = set(stopwords.words("english")) - negations
```

```python
def remove_stop_words(words):
    return [word for word in words if word not in stop_words]

Data = Data.apply(remove_stop_words)
Data
```

<p>
Stopwords were removed using NLTK's English stopword list. The negations
<b>no, nor, not</b> were deliberately kept so the meaning of the sentences is preserved.
This reduces noise while keeping the meaningful words for downstream NLP tasks.
</p>

<span style="color:green"> 
POS-aware LEMMATIZATION
</span>

```python
from nltk import pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
nltk.download("wordnet")
nltk.download("omw-1.4") 

lt = WordNetLemmatizer()
```

```python
def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

```python
def lemmatize_with_pos(words):
    tagged_words = pos_tag(words)

    return [
        lt.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_words
    ]

example = ["running", "better", "studies", "wolves"]

lemmatize_with_pos(example)
```

```python
Data = Data.apply(lemmatize_with_pos)
```

<span style="color:green"> 
UNIGRAM | BIGRAM | N-GRAM
</span>

```python
Data = Data.apply(lambda x: " ".join(x))
```

```python
type(Data)
```

```python
def n_gram_generation(data,n = 1):
    if type(data).__name__ == "Series":
        for sentence in data:
            words = sentence.split()

            for i in range(len(words)-n+1):
                gram = " ".join(words[i:i+n])
                print(gram)
    else:
        print("Only accepting Series")
```

```python
print("UNIGRAMS");  n_gram_generation(Data, n=1)
print("BIGRAMS");   n_gram_generation(Data, n=2)
print("TRIGRAMS");  n_gram_generation(Data, n=3)
```

<span style="color:green"> 
ONE HOT ENCODING
</span>

```python
from sklearn.preprocessing import MultiLabelBinarizer

encoder = MultiLabelBinarizer()
ohe_matrix = encoder.fit_transform(Data.apply(word_tokenize))
ohe_df = pd.DataFrame(
    ohe_matrix,
    columns=encoder.classes_
)
ohe_df
```

```python
from sklearn.preprocessing import OneHotEncoder

# One-hot encode every individual token
all_tokens = [tok for sent in Data.str.split() for tok in sent]
token_encoder = OneHotEncoder(sparse_output=False)
token_matrix = token_encoder.fit_transform(pd.DataFrame(all_tokens))
token_ohe_df = pd.DataFrame(token_matrix, columns=token_encoder.categories_[0], index=all_tokens)
token_ohe_df.head(10)
```
