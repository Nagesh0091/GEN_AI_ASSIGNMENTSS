```python
from dotenv import load_dotenv
from google import genai

load_dotenv() # LOAD THE INFORMATION INTO PYTHON CODE 
client = genai.Client()  # CREATES A GEMINI CLIENT USING THE API KEY FROM ENVIRONMENT VARIABLE
```

```python
# LIST OF GEMINI MODEL TO TRY IF ONE MODEL FAILS
MODELS = ["gemini-3.8-flash", "gemini-3.5-flash-lite", "gemini-3.1-flash-lite"]
```

```python
def ask(prompt):
    # TRY EACH MODEL ONE BY ONE
    for model in MODELS:
        try:
            # SEND THE PROMPT TO THE SELECTED GEMINI MODEL
            response = client.models.generate_content(model=model, contents=prompt)

            # RETURN THE GENERATED RESPONSE
            return response.text

        except:
            # IF THE CURRRENT MODEL FAILS TRY THE NEXT MODEL
            print(model, "failed")

    # RETURN THIS MESSAGE IF ALL MODELS FAILED
    return "All models failed"
```

```python
#EXPLAINER
prompt_1 = """Explain the main difference between a GAN (Generative Adversarial Network)
and a VAE (Variational Autoencoder). Cover how each one learns and how each one
generates new data. Keep the answer short and clear, in about 5 sentences,
suitable for a student who just attended an introductory lecture on Generative AI."""

response_1 = ask(prompt_1)
print(response_1)
```

```python
#POET
prompt_2 = """Write a short poem about artificial intelligence. The poem must have exactly four lines. Return only the poem."""
response_2 = ask(prompt_2)
print(response_2)
```

```python
#CODER
prompt_3 = """Write a simple Python function called greet that takes a name as input and prints a personalized greeting, for example "Hello, Asha! Welcome!". Include a short comment and one example call. Return only the Python code."""
response_3 = ask(prompt_3)
print(response_3)
```

```python
#PERSONA
prompt_4 = """Act like a friendly pirate. Speak in pirate style (use words like "Ahoy", "matey" and "arr") and explain what Generative AI is in 4 to 5 sentences."""
response_4 = ask(prompt_4)
print(response_4)
```

```python
reflection = """Working with Gemini showed me how much the wording of a prompt changes the answer.
When I added limits such as "exactly four lines" or "about 5 sentences", the output matched
what I wanted much more closely. The persona task was the most fun, because the same topic
sounded completely different in a pirate voice. I also learned that API keys must be kept
in a .env file and never written in the code. Next time I would like to try changing the
temperature to see how it affects creativity."""

print(reflection)
```

```python
# CONSOLIDATED IN Assignment7_Gemini_Output.md
tasks = [("Explainer", prompt_1, response_1),
         ("Poet", prompt_2, response_2),
         ("Coder", prompt_3, response_3),
         ("Persona", prompt_4, response_4)]

# CREATE AND SAVE THE OUTPUT IN MARKDOWN FILE
with open("Assignment7_Gemini_Output.md", "w", encoding="utf-8") as f:
    for title, prompt, response in tasks:
        f.write(f"## {title}\n\n")
        f.write(f"**Prompt:**\n\n{prompt}\n\n")
        f.write(f"**Model Output:**\n\n{response}\n\n")
    f.write("## Personal Reflection\n\n")
    f.write(reflection)

print("PROMT AND RESPONSES ARE SAVED SUCCESSFULLY")
```
