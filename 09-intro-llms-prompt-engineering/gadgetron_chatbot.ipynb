````python
from dotenv import load_dotenv
from google import genai

load_dotenv() # LOAD THE INFORMATION INTO PYTHON CODE 
client = genai.Client()  # CREATES A GEMINI CLIENT USING THE API KEY FROM ENVIRONMENT VARIABLE
````

````python
# LIST OF GEMINI MODEL TO TRY IF ONE MODEL FAILS
MODELS = ["gemini-3.5-flash-lite", "gemini-3.1-flash-lite"]
````

````python
from google.genai import types

# CREATE A FUNCTION TO SEND PROMPTS TO GEMINI
def ask(prompt, temperature=None):

    # SET THE TEMPERATURE IF PROVIDED
    config = types.GenerateContentConfig(temperature=temperature) if temperature is not None else None

    # STORE THE LAST ERROR
    last_error = None

    # TRY EACH GEMINI MODEL ONE BY ONE
    for model in MODELS:
        try:
            # SEND THE PROMPT TO THE GEMINI MODEL
            response = client.models.generate_content(
                model=model,
                contents=prompt,
                config=config
            )

            # RETURN THE GENERATED RESPONSE
            return response.text

        except Exception as e:
            # PRINT THE ERROR IF THE CURRENT MODEL FAILS
            print(f"[{model}] failed: {str(e)[:120]}")

            # STORE THE ERROR
            last_error = e

    # RAISE THE ERROR IF ALL MODELS FAIL
    raise last_error
````

````python
# CREATE A FEW-SHOT PROMPT FOR TICKET CLASSIFICATION
prompt1 = """You are a support assistant for "Gadgetron", an online electronics store.

Classify each customer email into exactly ONE department: Shipping, Billing, or Returns.
Reply with only the department name.

# EXAMPLE 1: SHIPPING
Email: "My tracking page says delivered, but I never received my package. Where is it?"
Department: Shipping

# EXAMPLE 2: BILLING
Email: "I was charged twice for the same order. Please refund the extra charge."
Department: Billing

# EXAMPLE 3: RETURNS
Email: "The jacket size is wrong and I would like to send it back for a refund."
Department: Returns

# CLASSIFY THE NEW CUSTOMER EMAIL
Email: "Hi, I received my order (#GT-58291) today, but the headphones inside the box are broken. How do I send them back for a replacement?"
Department:"""

# SEND THE PROMPT TO GEMINI
response1 = ask(prompt1, temperature=0)

# PRINT THE CLASSIFICATION RESULT
print(response1)
````

````python
# CREATE A PROMPT FOR A STANDARD AND PROFESSIONAL APOLOGY
prompt2a = """You are a customer support agent for "Gadgetron", an online electronics store.
Write a standard and professional apology to a customer whose order arrived late.
Keep it to 3 to 4 sentences, use a polite and formal tone, and do not invent
specific details such as dates, names or order numbers."""

# GENERATE A RESPONSE USING LOW TEMPERATURE (0.2)
response2a = ask(prompt2a, temperature=0.2)

# PRINT THE GENERATED APOLOGY
print(response2a)
````

````python
# CREATE A PROMPT FOR AN EMPATHETIC AND CREATIVE APOLOGY
prompt2b = """You are a customer support agent for "Gadgetron", an online electronics store.
Write an empathetic and creative apology to a customer whose order arrived late.
Make it feel warm and personal, show that you understand how frustrating waiting can be,
and use fresh, original wording instead of stock phrases. Keep it to 3 to 4 sentences
and do not invent specific details such as dates, names or order numbers."""

# GENERATE A RESPONSE USING HIGH TEMPERATURE (0.9)
response2b = ask(prompt2b, temperature=0.9)

# PRINT THE GENERATED APOLOGY
print(response2b)
````

````python
# CREATE A PROMPT USING THE COMPANY'S RETURN POLICY AS CONTEXT
prompt3 = """You are a customer support chatbot for "Gadgetron", an online electronics store.
Answer the customer's question using ONLY the return policy below.
Do not use outside knowledge and do not make assumptions.
If the answer is not in the policy, reply exactly: "I do not have that information."
Answer every part of the customer's question.

# COMPANY'S OFFICIAL RETURN POLICY
Return Policy:
"Customers can return any unused item in its original packaging within 30 days of purchase for a full refund. The customer is responsible for return shipping costs. Sale items are non-refundable."

# CUSTOMER'S QUESTION
Customer question: "Can I return a laptop I bought on sale last week? And will you pay for the shipping?"

Answer:"""

# SEND THE PROMPT TO GEMINI
response3 = ask(prompt3, temperature=0)

# PRINT THE GENERATED RESPONSE
print(response3)
````

````python
# STORE ALL TASKS, PROMPTS AND GENERATED RESPONSES
tasks = [("Ticket Routing (Few-Shot)", prompt1, response1),
         ("Standard Apology (Temperature 0.2)", prompt2a, response2a),
         ("Creative Apology (Temperature 0.9)", prompt2b, response2b),
         ("Answer with Context", prompt3, response3),]

# CREATE AND OPEN A MARKDOWN FILE
with open("Assignment9_Output.md", "w", encoding="utf-8") as f:

    # WRITE THE MAIN HEADING
    f.write("# Assignment 9: Gadgetron Chatbot\n\n")

    # WRITE EACH TASK, PROMPT AND RESPONSE
    for title, prompt, response in tasks:

        # WRITE THE TASK TITLE AND PROMPT
        f.write(f"## {title}\n\n**Prompt:**\n\n```\n{prompt}\n```\n\n")

        # WRITE THE GENERATED RESPONSE
        f.write(f"**Response:**\n\n{response}\n\n---\n\n")

# DISPLAY THE SUCCESS MESSAGE
print("Saved Assignment9_Output.md")
````
