```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()  # reads GEMINI_API_KEY from the .env file

# Current Gemini models, tried in order if one is unavailable
MODELS = ["gemini-3.8-flash", "gemini-3.5-flash-lite", "gemini-3.1-flash-lite"]

def ask(prompt):
    last_error = None
    for model in MODELS:
        try:
            response = client.models.generate_content(
                model=model,
                contents=prompt
            )
            return response.text
        except Exception as e:
            print(f"[{model}] failed: {str(e)[:120]}")
            last_error = e
    raise last_error
```

```python
# DATA EXTRACTING
prompt1 = """Read the text below and do two things:
1. Summarize it in a single sentence.
2. Extract the key details into a valid JSON object with the keys: product_name, release_date, screen_size, weight, base_ram, base_price, main_complaint.

Text:
"The new 'Starlight' laptop was released on October 26, 2024. It features a 14-inch OLED screen, weighs 2.5 lbs, and the base model with 16GB of RAM costs $1,299. The main complaint from early reviews is its short 6-hour battery life."
"""

response1 = ask(prompt1)
print(response1)
```

```python
# CODE GENERATING
prompt2 = """Write a Python function called calculate_average that takes a list of numbers as input and returns the average of those numbers. Handle an empty list by returning 0. Return only the code."""

response2 = ask(prompt2)
print(response2)
```

```python
# CODE TRANSLATING 
prompt3 = f"""Convert the following Python code into JavaScript. Keep the same logic and function name. Return only the JavaScript code.

{response2}
"""

response3 = ask(prompt3)
print(response3)
```

```python
# SOCIAL MEDIA MANAGER
prompt4 = """Act as a witty and enthusiastic social media manager. Write a tweet announcing the 'Starlight' laptop, a 14-inch OLED laptop that weighs 2.5 lbs and starts at $1,299. The tweet must include exactly two emojis and the hashtag #TechLaunch."""

response4 = ask(prompt4)
print(response4)
```

```python
tasks = [
    ("Task 1: The Data Extractor", prompt1, response1),
    ("Task 2: The Code Generator", prompt2, response2),
    ("Task 3: The Code Translator", prompt3, response3),
    ("Task 4: The Social Media Manager", prompt4, response4),
]

with open("AI_Prompting_Showcase.md", "w", encoding="utf-8") as f:
    for title, prompt, response in tasks:
        f.write(f"## {title}\n\n")
        f.write(f"**Prompt:**\n\n{prompt}\n\n")
        f.write(f"**Response:**\n\n{response}\n\n---\n\n")

print("Saved AI_Prompting_Showcase.md")
```
